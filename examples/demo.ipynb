# ctrl_viz Demo Notebook

This notebook demonstrates how to use the ctrl_viz library for creating Bode and Nyquist plots of control systems.

```python
# Import required libraries
import numpy as np
import control
from ctrl_viz import bode_plot, bode_magnitude, bode_phase, nyquist_plot
```

## 1. Define Transfer Functions

Let's create some example transfer functions to visualize.

```python
# First-order system: G(s) = 1 / (s + 1)
first_order = control.TransferFunction([1], [1, 1])
print("First-order system:")
print(first_order)
```

```python
# Second-order system: G(s) = 1 / (s^2 + 0.5s + 1)
second_order = control.TransferFunction([1], [1, 0.5, 1])
print("Second-order system:")
print(second_order)
```

```python
# Third-order system with gain: G(s) = 10 / ((s+1)(s+2)(s+10))
third_order = control.TransferFunction([10], np.convolve(np.convolve([1, 1], [1, 2]), [1, 10]))
print("Third-order system:")
print(third_order)
```

## 2. Bode Plots

### 2.1 Full Bode Plot (Magnitude and Phase)

```python
# Bode plot for first-order system
fig, axes = bode_plot(first_order, title="First-Order System Bode Plot")
```

```python
# Bode plot for second-order system
fig, axes = bode_plot(second_order, title="Second-Order System Bode Plot")
```

```python
# Bode plot with custom frequency range
omega = np.logspace(-3, 3, 1000)
fig, axes = bode_plot(third_order, omega=omega, title="Third-Order System Bode Plot")
```

### 2.2 Magnitude-Only Bode Plot

```python
# Magnitude plot for second-order system
fig, ax = bode_magnitude(second_order, title="Second-Order System Magnitude")
```

### 2.3 Phase-Only Bode Plot

```python
# Phase plot for second-order system
fig, ax = bode_phase(second_order, title="Second-Order System Phase")
```

### 2.4 Bode Plot Options

```python
# Bode plot with frequency in Hz
fig, axes = bode_plot(second_order, Hz=True, title="Bode Plot with Hz")
```

```python
# Bode plot with linear magnitude (not dB)
fig, axes = bode_plot(second_order, dB=False, title="Bode Plot with Linear Magnitude")
```

## 3. Nyquist Plots

```python
# Nyquist plot for first-order system
fig, ax = nyquist_plot(first_order, title="First-Order System Nyquist Plot")
```

```python
# Nyquist plot for second-order system
fig, ax = nyquist_plot(second_order, title="Second-Order System Nyquist Plot")
```

```python
# Nyquist plot with custom frequency range
omega = np.logspace(-3, 3, 2000)
fig, ax = nyquist_plot(third_order, omega=omega, title="Third-Order System Nyquist Plot")
```

### 3.1 Nyquist Plot Options

```python
# Nyquist plot without unit circle
fig, ax = nyquist_plot(second_order, unit_circle=False, title="Nyquist Plot without Unit Circle")
```

## 4. Practical Example: PID Controller Design

Let's look at a plant and its closed-loop system with a simple controller.

```python
# Plant: G(s) = 1 / (s(s+1)(s+5))
plant = control.TransferFunction([1], np.convolve(np.convolve([1, 0], [1, 1]), [1, 5]))
print("Plant:")
print(plant)
```

```python
# Bode plot of the plant
omega = np.logspace(-2, 2, 1000)
fig, axes = bode_plot(plant, omega=omega, title="Plant Bode Plot")
```

```python
# Nyquist plot of the plant
fig, ax = nyquist_plot(plant, omega=omega, title="Plant Nyquist Plot")
```
